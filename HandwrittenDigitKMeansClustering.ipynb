```python
#Importing pandas and scikitlearn libraries
import matplotlib.pyplot as plt
import numpy as np
from sklearn.cluster import MiniBatchKMeans
```

Importing MNIST data. The training data has 60,000 images and testing data has 10,000 images)

```python
# Importing the dataset from keras
from keras.datasets import mnist
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

Clustering algorithms require 1-dimensional array.

MNIST dataset has images that are 28 by 28 pixels. Convert them into one dimensional array of length 784 (28x28) pixels for 60,000 images

```python
# Normalize pixel values of images by dividing by 255
x_train = x_train.astype('float32')
x_train = x_train/255.0

# Convert to 1-dimensional array
x_train = x_train.reshape(60000,28*28)
```

Create MiniBatchKMeans model (for more efficiency than KMeans) and apply model to data

```python
# Initialize the K-Means model for 10 clusters
kmeans = MiniBatchKMeans(n_clusters = 10, n_init=3)

# Fitting the model to training set
kmeans.fit(x_train)
kmeans.labels_
```

kmeans.labels_ gives us the cluster number for each of the 10 clusters. Using y_train, find the value of the digit that most represents each of these clusters.

```python
dictionary_labels = {}

# For loop to run through each label in cluster labels
for i in range(len(np.unique(kmeans.labels_))):
  index = np.where(kmeans.labels_ == i,1,0)
  num = np.bincount(y_train[index==1]).argmax()
  dictionary_labels[i] = num

# Look into one of the clusters, say cluster 8, for bin counts of values
i = 8
index = np.where(kmeans.labels_ == i,1,0)
print(np.bincount(y_train[index==1]))

#dictionary_labels is a dictionary object which has a mapping of cluster number and digit
#create an array digit_values similar format as y_train which makes comparison easier
digit_values = np.random.rand(len(kmeans.labels_))
for i in range(len(kmeans.labels_)):
  digit_values[i] = dictionary_labels[kmeans.labels_[i]]
```

Check predicted values

```python
# Comparing Predicted values and Actual values for 20 samples
print(digit_values[:20].astype('int'))
print(y_train[:20])
```

Calculate accuracy score

```python
# Calculating accuracy score
from sklearn.metrics import accuracy_score
print(accuracy_score(y_train, digit_values))
```

Cluster size & Accuracy - Try a few different clusters to see what happens to accuracy



```python
from sklearn import metrics
cluster_number = [10,20,40,80,120,250,400]

for i in cluster_number:
  # Initialize the K-Means model for cluster i
  kmeans = MiniBatchKMeans(n_clusters = i, n_init=3)
  # Fit the model to training set
  kmeans.fit(x_train)

  # For each cluster label, find digit values that most represent it
  dictionary_labels = {}
  for i in range(len(np.unique(kmeans.labels_))):
    index = np.where(kmeans.labels_ == i,1,0)
    num = np.bincount(y_train[index==1]).argmax()
    dictionary_labels[i] = num

  digit_values = np.random.rand(len(kmeans.labels_))
  for i in range(len(kmeans.labels_)):
    digit_values[i] = dictionary_labels[kmeans.labels_[i]]

  #Display number of clusters and accuracy score
  print('Clusters : {}'.format(kmeans.n_clusters))
  print('Accuracy : {}'.format(accuracy_score(y_train, digit_values)))
  print('\n')
```

Cluster Centroids

```python
# Extract centroids from kmeans and set it to the right format
centroids = kmeans.cluster_centers_
centroids.shape
centroids = centroids.reshape(400,28,28)

#We had normalized the data. So now we nullify the normalization effect by multiplying by 255.
centroids = centroids * 255

#visualize the cluster centroids. Create a figure with subplots for 32 images with 8 in each row for 4 rows
#Loop through subplots and add centroid images

plt.figure(figsize = (6,6))
for i in range(32):
 plt.subplot(4,8,i+1)
 plt.title('Digit:{}'.format(dictionary_labels[i]),fontsize = 8)
 plt.imshow(centroids[i])
```
